### Problem A3: Monte Carlo Simulation - Estimating Pi

One of the most powerful applications of loops and random numbers is
**Monte Carlo simulation** -- a computational technique used by engineers
to solve problems through random sampling.

#### Background: The Dart Board Method

Imagine throwing darts randomly at a 1x1 square with a quarter-circle
(radius = 1) in the corner:
- **Square area** = 1
- **Quarter-circle area** = π/4
- **Probability dart lands in quarter-circle** = π/4

Therefore: **π ≈ 4 x (darts in quarter-circle) / (total darts)**

A point (x, y) is inside the quarter-circle if: **x² + y² ≤ 1**

#### Problem B2.6: Order Dependence in Randomness

You know that setting a seed makes random numbers reproducible. But
reproducibility depends not only on the **seed** -- it also depends on the
**order of random calls**. Even small changes in control flow (loops,
conditionals) can completely change the output sequence.

**Instructions**
1. Import the `random` module and set the seed with `random.seed(123)`.
2. Run both Snippet A and Snippet B below and compare the outputs.

**Snippet A**
```python
import random
random.seed(123)

for i in range(3):
    x = random.randint(1, 10)
    if x % 2 == 0:
        y = random.random()
        print(f"Loop {i+1}: x={x}, y={y:.3f}")
    else:
        print(f"Loop {i+1}: x={x}")
```

**Snippet B**
```python
import random
random.seed(123)

for i in range(3):
    x = random.randint(1, 10)
    y = random.random()
    if x % 2 == 0:
        print(f"Loop {i+1}: x={x}, y={y:.3f}")
    else:
        print(f"Loop {i+1}: x={x}")
```

**Questions**
- Do Snippet A and Snippet B produce identical outputs? Why or why not?
- How does the placement of `random.random()` affect the sequence of
  random values consumed?
- If two students use the same seed but structure their code differently
  (e.g., conditionals or loop order), will they get the same results?
- What does this teach you about reproducibility in scientific code?

🚩 **Common mistake:** assuming the same seed always produces the same
*visible* output regardless of code structure. The seed fixes the
underlying sequence of values `random` will hand out, in order -- but
which calls in your code actually *consume* values from that sequence,
and in what order, depends entirely on your program's control flow.
