### Problem 4: Content Filter

Create a content filtering system that analyzes text. **This version uses
NO LOOPS** -- you'll use the techniques from Class 3.

**Budget extra time for this one -- it's the longest problem.** You'll
check each word individually with separate Boolean variables. Once you
learn loops, you'll see how to write this in about 10 lines instead of
100!

See the reading's [Comparing to Boolean Values](https://BU-EK125.github.io/EK125/class/Class3.html#comparing-to-boolean-values) section -- the starter code below deliberately writes comparisons like `(hasViolence == True)`, which is exactly the pattern that section discusses.


````python
# Get text input
userText = input("Enter text to analyze: ")
targetAudience = input("Target audience (child/teen/adult): ").lower()

# Basic text analysis using string methods
textLength = len(userText)
wordList = userText.split()
wordCount = len(wordList)
sentenceCount = userText.count(".") + userText.count("!") + userText.count("?")

# Text complexity analysis
hasLongText = (textLength > 500)
hasManyWords = (wordCount > 100)
hasManySentences = (sentenceCount > 10)

# Check for inappropriate content - NO LOOPS!
# We'll check each word individually using the technique from Class 3
lowerText = userText.lower()

hasViolence = ("violence" in lowerText)
hasHate = ("hate" in lowerText)
hasInappropriate = ("inappropriate" in lowerText)
hasBad = ("bad" in lowerText)

# Combine with OR (if ANY inappropriate word is found)
hasInappropriateContent = (hasViolence == True) or (hasHate == True) or (hasInappropriate == True) or (hasBad == True)

# Build list of found inappropriate words
foundInappropriateWords = []
if hasViolence == True:
    foundInappropriateWords.append("violence")
if hasHate == True:
    foundInappropriateWords.append("hate")
if hasInappropriate == True:
    foundInappropriateWords.append("inappropriate")
if hasBad == True:
    foundInappropriateWords.append("bad")

# Check for positive content - NO LOOPS!
hasGreat = ("great" in lowerText)
hasAwesome = ("awesome" in lowerText)
hasWonderful = ("wonderful" in lowerText)
hasExcellent = ("excellent" in lowerText)
hasAmazing = ("amazing" in lowerText)

# Combine with OR
hasPositiveContent = (hasGreat == True) or (hasAwesome == True) or (hasWonderful == True) or (hasExcellent == True) or (hasAmazing == True)

# Build list of found positive words
foundPositiveWords = []
if hasGreat == True:
    foundPositiveWords.append("great")
if hasAwesome == True:
    foundPositiveWords.append("awesome")
if hasWonderful == True:
    foundPositiveWords.append("wonderful")
if hasExcellent == True:
    foundPositiveWords.append("excellent")
if hasAmazing == True:
    foundPositiveWords.append("amazing")

# Age-appropriate analysis using explicit Boolean comparisons
isAppropriateForChildren = (hasInappropriateContent == False) and (textLength <= 200) and (hasManyWords == False)
isAppropriateForTeens = (hasInappropriateContent == False) and (textLength <= 1000)
isAppropriateForAdults = True  # Adults can read anything

# Content rating based on target audience
isContentApproved = False
if (targetAudience == "child") and (isAppropriateForChildren == True):
    isContentApproved = True
elif (targetAudience == "teen") and (isAppropriateForTeens == True):
    isContentApproved = True
elif targetAudience == "adult":
    isContentApproved = True

# Reading level analysis
if sentenceCount > 0:
    averageWordsPerSentence = wordCount / sentenceCount
else:
    averageWordsPerSentence = 0

print (f"is the content approved? {isContentApproved}")

# TODO: Classify reading level
# Easy reading: average 15 words or fewer per sentence
# Medium reading: average 16-25 words per sentence
# Difficult reading: average more than 25 words per sentence

# Your code here:


# TODO: Print content analysis report using explicit Boolean checks
# Expected output should show text statistics, approval status, found words, and reading level
#
# Example format #1:
# --- Text Analysis Report ---
# Text length: 40 characters
# Word count: 8
# APPROVED: Content approved for child audience
# READING LEVEL: Easy
#
# Example format #2:
# --- Text Analysis Report ---
# Text length: 292 characters
# Word count: 46
# APPROVED: Content approved for teen audience
# WARNING: Found inappropriate words: [\'bad\']
# POSITIVE: Found positive words: [\'great\', \'awesome\']
# READING LEVEL: Easy
#
# Example format #3:
# --- Text Analysis Report ---
# Text length: 450 characters
# Word count: 68
# APPROVED: Content approved for adult audience
# POSITIVE: Found positive words: [\'wonderful\']
# READING LEVEL: Difficult

# Your code here:
````
