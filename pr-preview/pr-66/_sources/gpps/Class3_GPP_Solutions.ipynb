## Part 2: Weather Decision System

Create a program that helps decide what to wear based on weather conditions.

**Your Task:** Complete the TODO sections with appropriate Boolean expressions using parentheses.

**Important:** Use explicit Boolean comparisons throughout!

```python
# Get weather conditions
temperature = float(input("Enter temperature (°F): "))
isRaining = input("Is it raining? (yes/no): ").lower() == "yes"
isWindy = input("Is it windy? (yes/no): ").lower() == "yes"
humidity = float(input("Enter humidity percentage: "))

# Decision logic - complete these conditions using explicit comparisons
needsJacket = (temperature < 60) or (isWindy == True)
needsUmbrella = isRaining == True
isComfortable = (isRaining == False) and (temperature >= 65) and (temperature <= 80) and (humidity < 70)
stayInside = (temperature < 0) or (temperature > 90) or ((isRaining == True) and (isWindy == True))

# Output recommendations using explicit Boolean checks
print("\n--- Weather Recommendations ---")
if needsJacket == True:
    print("Bring a jacket!")

if needsUmbrella == True:
    print("Take an umbrella!")

if isComfortable == True:
    print("Perfect weather to be outside!")

if stayInside == True:
    print("Maybe consider staying indoors today")
```

## Part 3: Grade Calculator with Conditions

Create a grade evaluation system using clear Boolean naming and explicit comparisons:

**Your Task:** Complete the Boolean expressions using proper naming and parentheses.

```python
# Student information
homeworkScore = float(input("Enter homework average (0-100): "))
examScore = float(input("Enter exam score (0-100): "))
attendanceRate = float(input("Enter attendance rate (0-100%): "))
hasExtraCredit = input("Extra credit completed? (yes/no): ").lower() == "yes"

# Grade calculations
finalScore = (homeworkScore * 0.4) + (examScore * 0.6)
if hasExtraCredit == True:
    finalScore = finalScore + 5  # 5 point bonus

# Conditions to determine - work together on these using good Boolean naming
isPassing = (finalScore >= 60) and (attendanceRate >= 75)
hasHonors = (finalScore >= 90) and (attendanceRate >= 95)
needsImprovement = ((homeworkScore < 70) or (examScore < 65)) and (attendanceRate < 80)
isExcellent = (finalScore >= 95) and (attendanceRate == 100) and (hasExtraCredit == True)

# Output results using explicit Boolean comparisons
print(f"\nFinal Score: {finalScore:.1f}")
print(f"Attendance: {attendanceRate}%")

if isExcellent == True:
    print("Excellent work! You're a star student!")
elif hasHonors == True:
    print("Honors achievement!")
elif isPassing == True:
    print("Passing grade - well done!")
else:
    print("Not passing - please see instructor")

if needsImprovement == True:
    print("Areas for improvement identified")
```

## Part 4: Login Validator (REVISED - No Loops!)

**Challenge:** Create a truth table on paper for the `canLogin` logic before testing your code!

**Note:** This problem uses string comparison tricks instead of loops. We'll learn loops in Class 5!

```python
# User credentials
username = input("Enter username: ")
password = input("Enter password: ")
isWeekend = input("Is it weekend? (yes/no): ").lower() == "yes"
hourOfDay = int(input("Enter hour (0-23): "))

# Password validation using string methods (no loops needed!)
# These methods check the ENTIRE string at once
hasMinLength = (len(password) >= 8)
hasUpperCase = (password != password.lower())  # If string changes when lowercased, it has uppercase!

# Check for digits by seeing if string contains any number
# This uses the 'in' operator multiple times combined with 'or'
hasDigit = ("0" in password) or ("1" in password) or ("2" in password) or ("3" in password) or ("4" in password) or ("5" in password) or ("6" in password) or ("7" in password) or ("8" in password) or ("9" in password)

# Validation rules - figure these out together using explicit Boolean logic
isValidUsername = (len(username) >= 5) and (username.isalnum() == True)
isValidPassword = (hasMinLength == True) and (hasDigit == True) and (hasUpperCase == True)
isBusinessHours = (hourOfDay >= 9) and (hourOfDay <= 17) and (isWeekend == False)
hasAdminAccess = (username.lower() == "admin") and (password == "Admin123")

# Access decision using explicit Boolean checks
canLogin = ((isValidUsername == True) and (isValidPassword == True)) or (hasAdminAccess == True)
fullAccess = (canLogin == True) and (isBusinessHours == True)

print(f"\nUsername valid: {isValidUsername}")
print(f"Password valid: {isValidPassword}")
print(f"Business hours: {isBusinessHours}")

if fullAccess == True:
    print("Full access granted!")
elif canLogin == True:
    print("Limited access granted (outside business hours)")
else:
    print("Access denied")
```
