### Problem 2.3: While Loops — Counting and Averages

Ahead of time, we do not know how many times the action of a `while` loop will be executed. However, it is frequently useful to **count** the number of times that the action has been executed.

Write a script that:  
- Prompts the user to enter **positive numbers**, one at a time.  
- Continues until the user enters a **negative number**.  
- Keeps track of both:  
  - The **sum** of the positive numbers entered.  
  - The **count** of how many numbers were entered.  
- At the end:  
  - If no positive numbers were entered, print an **error message**.  
  - Otherwise, print the **average** of the positive numbers.  
  - Also print how many tries it took, in the form:  
    ```
    It took xx tries.
    ```
  - If the user stops the loop after the very first input, print:  
    ```
    It took 1 try.
    ```


````python
# Initialize variables
total = 0
count = 0

# Prompt for the first number
num = float(input("Enter a positive number (or a negative number to stop): "))

# Loop while the number is positive
while num >= 0:
    total += num
    count += 1
    num = float(input("Enter a positive number (or a negative number to stop): "))

# After the loop
if count == 0:
    print("Error: no positive numbers were entered.")
else:
    average = total / count
    print(f"The average of the {count} numbers is {average:.2f}")
    if count == 1:
        print("It took 1 try.")
    else:
        print(f"It took {count} tries.")
````

### Problem 2.5 Error checking, modules and more!

Write code that will present the user with choices for functions:  

- **1 for doubling the number**  
- **2 for squaring the number**  
- **3 for inverting the number (1/x)**  

Error-check by looping to display the choices until the user enters 1, 2, or 3. Then, generate a **random real number** between 0 and 1, and print the result of the user’s chosen function on that number.

**Hint**: To generate random *real numbers* between 0 and 1, you will need a special function from the `random` library, which is confusingly *also called* `random()`. Be sure to import the `random()` from from the `random` library using the same synatx as abvoe. Use `help()` to get some information about how `random()` works!

You will see something like this

```python

help(random)
```
```
Help on built-in function random:

random() method of random.Random instance
    random() -> x in the interval [0, 1).
```

````python
from random import random

print("Please choose a function:")
print("1 for double it")
print("2 for square it")
print("3 for invert it (1/x)")

mychoice = int(input('Enter your choice: '))
while mychoice < 1 or mychoice > 3:
    print('The choice must be 1, 2, or 3.')
    mychoice = int(input('Enter your choice: '))

x = random()

if mychoice == 1:
    result = 2 * x
    print(f'Doubling {x:.4f} gives {result:.4f}')

elif mychoice == 2:
    result = x * x
    print(f'Squaring {x:.4f} gives {result:.4f}')

else:
    if x == 0:
        print("Oops, can't invert zero!")
    else:
        result = 1 / x
        print(f'Inverting {x:.4f} gives {result:.4f}')
````
