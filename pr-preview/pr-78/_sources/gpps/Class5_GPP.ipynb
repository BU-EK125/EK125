## Part 2: `while` loops

### Problem 2.1: A `while` loop that might not run

Here is an example `while` loop:

```python
# Example: a loop that runs
count = 5
while count > 0:
    print("Counting down:", count)
    count = count - 1
print("Done!")
```

When you run this code, you should see:

```
Counting down: 5
Counting down: 4
Counting down: 3
Counting down: 2
Counting down: 1
Done!
```

In this example, the condition `count > 0` is **True** when the loop starts, so the body of the loop runs several times before stopping.

- Write code with an initial value for `count` that makes the loop body run **exactly once**?  
- Write code with an initial value for `count` that makes the loop body **never run at all**?  

🚩 **Common mistake:** forgetting to update the loop-control variable
inside a `while` loop's body. Every `while` loop you write from here on
depends on something changing inside the loop so the condition
eventually becomes false -- skip that update and the loop never ends.
See the reading's [While and while-else](https://BU-EK125.github.io/EK125/class/Class5.html#while-and-while-else) section.
